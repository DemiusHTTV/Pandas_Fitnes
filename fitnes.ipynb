5. Получи датафрейм с разбивкой выручки по месяцам и по типам тренировок с помощью группировки с суммой методом `groupby()`.

```python
payments['payment_date'] = pd.to_datetime(payments['payment_date'], dayfirst=True)
```

```python
USD_RATE = 73.14 # курс доллара

def map_currency(currency): # определи функцию для заполнения столбца fx_rate методом .apply()
 if currency == "USD":
    return USD_RATE
 else:
    return 1
    

payments_2020 = payments[payments['payment_date'].dt.year == 2020].copy()

payments_2020['fx_rate'] = payments_2020['currency'].apply(map_currency)
print(payments_2020)
```

```python
payments_2020['amount_rub'] = payments_2020['amount'] * payments_2020['fx_rate']
payments_2020.head()
```

```python
payments_2020['payment_month'] = payments_2020['payment_date'].dt.month
revenue_month = payments_2020.groupby(['payment_month', 'workout_type'])['amount_rub'].sum().reset_index()
print(revenue_month)
```
